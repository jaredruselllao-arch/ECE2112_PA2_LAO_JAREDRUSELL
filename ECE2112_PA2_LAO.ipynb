A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
import numpy as np 
```

```python
RAG=np.random.seed(2112)#Random array Generator
RAG
```

```python
X = np.random.randint(10, 101, size=(5, 5))
```

```python
X #Random Number Generator
```

```python
X_normalized=((X-np.mean(X))/np.std(X))
```

```python
X_normalized #The formula for getting X_normalized
```

```python
standard_deviation = np.std(X_normalized)
```

```python
standard_deviation #standard deviation of X_normalized
```

```python
mean = np.mean(X_normalized)
```

```python
mean#mean of X_normalized
```

```python
np.save("X_normalized.npy",X_normalized)
```

B. CUBES DIVISIBLE BY 4 PROBLEM

```python
C = (np.arange(1,101)**3).reshape(10, 10)
C
```

```python
C.shape #The shape of the array
```

```python
div_by_4 = C[C%4==0]
div_by_4
```

```python
div_by_4.size #Number of elements in the array that is divisible by four 
```

```python
np.save("div_by_4.npy", div_by_4)
```

C. ABOVE-MEAN SQUARES PROBLEM

```python
S = (np.arange(1,37)**2).reshape(6,6)
```

```python
S #Display of the 6x6 array of the first positive 36 integers all squared
```

```python
S_mean = np.mean(S)
```

```python
S_mean #The mean of the array S
```

```python
above_mean = S[S>S_mean]
```

```python
above_mean #Displaying all numbers in the array that are above the mean
```

```python
above_mean.size #counts the amount of elements above the mean
```

```python
np.save("above_mean.npy", above_mean)
```
